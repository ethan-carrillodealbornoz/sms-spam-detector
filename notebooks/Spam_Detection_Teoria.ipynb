# Clasificación de Spam con Naive Bayes

El spam o correo no solicitado representa un problema constante en las comunicaciones digitales. Además de saturar las bandejas de entrada, supone un riesgo de seguridad al ser utilizado como vector en ataques de _phishing_ y distribución de malware. La detección efectiva de spam es fundamental para mantener la integridad y operatividad de los sistemas de mensajería.

## Teorema de Bayes

El Teorema de Bayes es un principio fundamental en la teoría de la probabilidad que describe la probabilidad de que ocurra un evento basándose en el conocimiento previo de las condiciones asociadas a dicho evento.

Matemáticamente se expresa como:

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$

Donde:

- $P(A \mid B)$: Probabilidad a posteriori (_posterior probability_) de que ocurra el evento $A$ dado que $B$ es verdadero.
    
- $P(B \mid A)$: Verosimilitud (_likelihood_) de observar el evento $B$ dado que $A$ es verdadero.
    
- $P(A)$: Probabilidad a priori (_prior probability_) del evento $A$.
    
- $P(B)$: Verosimilitud marginal o evidencia (_marginal likelihood_) del evento $B$.


## Aplicación a la Detección de Spam

En el contexto del filtrado de correo, la hipótesis $A$ representa que un mensaje es spam ($\text{Spam}$), mientras que $B$ representa el conjunto de características observadas en el mensaje ($\text{Características}$), como palabras clave o cabeceras:

$$P(\text{Spam} \mid \text{Características}) = \frac{P(\text{Características} \mid \text{Spam}) \cdot P(\text{Spam})}{P(\text{Características})}$$

Desglose de los componentes:

1. **Probabilidad a Posteriori (**$P(\text{Spam} \mid \text{Características})$**):** Probabilidad de que un correo sea spam dadas sus características observadas.
    
2. **Verosimilitud (**$P(\text{Características} \mid \text{Spam})$**):** Probabilidad de que aparezcan dichas características en un correo que se sabe que es spam.
    
3. **Probabilidad a Priori (**$P(\text{Spam})$**):** Probabilidad general de que cualquier correo del sistema sea spam, independientemente de sus características.
    
4. **Verosimilitud Marginal (**$P(\text{Características})$**):** Probabilidad total de observar esas características en el conjunto global de correos (tanto spam como legítimos).
    

## La Asunción "Naive" (Ingenua)

El algoritmo Naive Bayes asume la hipótesis de independencia condicional: la presencia o ausencia de una característica determinada en un mensaje es completamente independiente de la presencia o ausencia de cualquier otra característica, dada la clase del mensaje.

Esta asunción simplifica el cálculo de la verosimilitud de un vector de $n$ características:

$$P(F_1, F_2, \dots, F_n \mid \text{Spam}) = \prod_{i=1}^{n} P(F_i \mid \text{Spam})$$

De igual forma, para la clase no spam ($\text{No Spam}$):

$$P(F_1, F_2, \dots, F_n \mid \text{No Spam}) = \prod_{i=1}^{n} P(F_i \mid \text{No Spam})$$

El modelo calcula la probabilidad a posteriori para ambas clases y asigna el mensaje a la clase con mayor probabilidad.

## Ejemplo Práctico de Cálculo

Dado un correo electrónico que presenta dos características $F_1$ y $F_2$, se desea determinar si debe clasificarse como spam.

### Datos de Entrada

- $P(\text{Spam}) = 0.3$: Probabilidad a priori de spam.
    
- $P(\text{No Spam}) = 0.7$: Probabilidad a priori de correo legítimo.
    
- $P(F_1 \mid \text{Spam}) = 0.4$: Probabilidad de la característica $F_1$ en spam.
    
- $P(F_2 \mid \text{Spam}) = 0.5$: Probabilidad de la característica $F_2$ en spam.
    
- $P(F_1 \mid \text{No Spam}) = 0.2$: Probabilidad de la característica $F_1$ en correo legítimo.
    
- $P(F_2 \mid \text{No Spam}) = 0.3$: Probabilidad de la característica $F_2$ en correo legítimo.
    

### 1. Cálculo de Verosimilitudes Combinadas

Aplicando la asunción de independencia:

$$P(F_1, F_2 \mid \text{Spam}) = P(F_1 \mid \text{Spam}) \cdot P(F_2 \mid \text{Spam}) = 0.4 \times 0.5 = 0.20$$$$P(F_1, F_2 \mid \text{No Spam}) = P(F_1 \mid \text{No Spam}) \cdot P(F_2 \mid \text{No Spam}) = 0.2 \times 0.3 = 0.06$$

### 2. Cálculo de la Verosimilitud Marginal

Utilizando la ley de probabilidad total:

$$P(F_1, F_2) = P(F_1, F_2 \mid \text{Spam}) \cdot P(\text{Spam}) + P(F_1, F_2 \mid \text{No Spam}) \cdot P(\text{No Spam})$$$$P(F_1, F_2) = (0.20 \times 0.3) + (0.06 \times 0.7) = 0.06 + 0.042 = 0.102$$

### 3. Cálculo de las Probabilidades a Posteriori

Para la clase **Spam**:

$$P(\text{Spam} \mid F_1, F_2) = \frac{P(F_1, F_2 \mid \text{Spam}) \cdot P(\text{Spam})}{P(F_1, F_2)} = \frac{0.06}{0.102} \approx 0.588 \quad (58.8\%)$$

Para la clase **No Spam**:

$$P(\text{No Spam} \mid F_1, F_2) = \frac{P(F_1, F_2 \mid \text{No Spam}) \cdot P(\text{No Spam})}{P(F_1, F_2)} = \frac{0.042}{0.102} \approx 0.412 \quad (41.2\%)$$

### Conclusión del Clasificador

Dado que $P(\text{Spam} \mid F_1, F_2) > P(\text{No Spam} \mid F_1, F_2)$ ($0.588 > 0.412$), el clasificador asigna el mensaje a la categoría **Spam**.

# Preprocesamiento de Texto para Clasificación de Spam

El preprocesamiento de texto estandariza el contenido, reduce el ruido y extrae características significativas para optimizar el rendimiento del clasificador Naive Bayes. Este pipeline utiliza la librería `nltk` para tareas de tokenización, eliminación de palabras vacías (_stop words_) y derivación regresiva (_stemming_).

## Descarga de Recursos de NLTK

Antes del procesamiento, es necesario descargar los paquetes de datos requeridos por `nltk`:

- `punkt` y `punkt_tab`: Modelos para la tokenización de oraciones y palabras.
    
- `stopwords`: Corpus de palabras comunes sin carga semántica decisiva.
    


```python
import nltk

# Descargar datos necesarios de NLTK
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

print("=== ESTADO INICIAL ===")
print(df.head(5))
```

## 1. Conversión a Minúsculas (_Lowercasing_)

Garantiza que el clasificador trate los términos por igual sin importar su capitalización original (por ejemplo, "Free" y "free" se procesan como el mismo token). Esto reduce la dimensionalidad del vocabulario y mejora la consistencia.

```python
# Convertir todo el texto a minúsculas
df["message"] = df["message"].str.lower()

print("\n=== TRAS CONVERSIÓN A MINÚSCULAS ===")
print(df["message"].head(5))
```

## 2. Limpieza de Puntuación y Números (Preservando Símbolos Clave)

Aunque la eliminación de caracteres especiales simplifica el dataset, ciertos símbolos como `$` y `!` aportan un contexto crítico en mensajes de spam (los precios o montos monetarios suelen incluir `$` y las llamadas de atención usan `!`).

Mediante expresiones regulares se eliminan dígitos y caracteres de puntuación generales, manteniendo únicamente letras en minúscula, espacios, `$` y `!`:

```python
import re

# Eliminar caracteres no deseados excepto letras, espacios, $ y !
df["message"] = df["message"].apply(lambda x: re.sub(r"[^a-z\s$!]", "", x))

print("\n=== TRAS ELIMINACIÓN DE PUNTUACIÓN Y NÚMEROS ===")
print(df["message"].head(5))
```

## 3. Tokenización (_Tokenization_)

La tokenización divide la cadena de texto en unidades individuales o palabras (_tokens_). Esta estructura permite la posterior manipulación término a término.


```python
from nltk.tokenize import word_tokenize

# Dividir el mensaje en lista de tokens
df["message"] = df["message"].apply(word_tokenize)

print("\n=== TRAS TOKENIZACIÓN ===")
print(df["message"].head(5))
```

## 4. Eliminación de Palabras Vacías (_Stop Words_)

Las _stop words_ son términos de alta frecuencia (como _and_, _the_, _is_) que no aportan valor discriminatorio entre clases. Su eliminación reduce el tamaño del conjunto de datos y enfoca el modelo en los términos informativos.

```python
from nltk.corpus import stopwords

# Cargar stop words en inglés y filtrarlas
stop_words = set(stopwords.words("english"))
df["message"] = df["message"].apply(lambda x: [word for word in x if word not in stop_words])

print("\n=== TRAS ELIMINACIÓN DE STOP WORDS ===")
print(df["message"].head(5))
```

## 5. Derivación Regresiva (_Stemming_)

El _stemming_ reduce cada palabra a su raíz morfológica (por ejemplo, _running_ a _run_). Consolida variantes de una misma palabra, reduciendo el tamaño total del vocabulario y mejorando la capacidad de generalización del clasificador.

Se utiliza el algoritmo **PorterStemmer**:

```python
from nltk.stem import PorterStemmer

# Aplicar el algoritmo Porter Stemmer a cada token
stemmer = PorterStemmer()
df["message"] = df["message"].apply(lambda x: [stemmer.stem(word) for word in x])

print("\n=== TRAS STEMMING ===")
print(df["message"].head(5))
```

## 6. Recomposición de Tokens a Cadenas de Texto

Aunque las listas de tokens son útiles para transformaciones individuales, la mayoría de los modelos de vectorización (como TF-IDF o Bag of Words) requieren cadenas de texto completas como entrada.

```python
# Unir los tokens procesados en un solo string separado por espacios
df["message"] = df["message"].apply(lambda x: " ".join(x))

print("\n=== TRAS RECOMPOSICIÓN A STRING ===")
print(df["message"].head(5))
```

Con este paso concluye la fase de preprocesamiento de texto. El dataset queda listo para la vectorización de características y el entrenamiento del clasificador Naive Bayes.

# Extracción de Características de Texto

La extracción de características transforma los mensajes de texto preprocesados en vectores numéricos legibles por los algoritmos de Machine Learning. Dado que los modelos matemáticos no pueden procesar texto en bruto de manera directa, requieren representaciones numéricas —como frecuencias o recuentos de palabras— para identificar patrones que permitan diferenciar el _spam_ del correo legítimo (_ham_).

## Representación de Texto como Vectores Numéricos

### El Modelo Bag-of-Words (Bolsa de Palabras)

El método más común para la representación vectorial de texto es el modelo **Bag-of-Words (BoW)**. Este enfoque:

1. Construye un vocabulario con todos los términos únicos presentes en el corpus.
    
2. Representa cada documento como un vector numérico donde cada posición corresponde a un término del vocabulario y su valor indica la frecuencia de aparición del término en dicho documento.
    

### Incorporación de Contexto Local: Unigramas y Bigramas

- **Unigramas (**$n=1$**):** Palabras individuales consideradas de forma independiente. No preservan el orden del texto ni la estructura sintáctica.
    
- **Bigramas (**$n=2$**):** Pares de palabras consecutivas. Permiten capturar contexto local e intenciones compuestas.
    

Por ejemplo, la secuencia o bigrama `free prize` aporta una señal mucho más clara de _spam_ que la palabra `free` aislada. Aunque los $n$-gramas introducen cierta relación sintáctica local, el orden global del texto sigue sin preservarse por completo.

## Uso de `CountVectorizer` en Scikit-Learn

La clase `CountVectorizer` de Scikit-Learn convierte una colección de documentos de texto en una matriz dispersa de recuentos de términos.

### Parámetros Principales

- `min_df`: Frecuencia mínima de documento. Un término debe aparecer en al menos $N$ documentos para incluirse en el vocabulario (evita palabras extremadamente raras o errores tipográficos).
    
- `max_df`: Frecuencia máxima de documento. Si un término aparece en más del porcentaje especificado de los documentos (por ejemplo, $0.90$ o $90\%$), se descarta por considerarse hiperfrecuente o poco discriminativo.
    
- `ngram_range`: Tupla $(min\_n, max\_n)$ que define el rango de $n$-gramas a extraer. Un valor de `(1, 2)` extrae tanto unigramas como bigramas.
    

### Implementación en Python

```python
from sklearn.feature_extraction.text import CountVectorizer

# Inicializar CountVectorizer con unigramas y bigramas, y filtrado por frecuencia
vectorizer = CountVectorizer(min_df=1, max_df=0.9, ngram_range=(1, 2))

# Ajustar el vocabulario y transformar los mensajes a matriz de frecuencias
X = vectorizer.fit_transform(df["message"])

# Etiqueta objetivo (target variable): 1 para spam, 0 para ham
y = df["label"].apply(lambda x: 1 if x == "spam" else 0)
```

## Etapas de Funcionamiento Interno

1. **Tokenización:** Segmentación del texto en función del rango de $n$-gramas configurado (`ngram_range`).
    
2. **Construcción del Vocabulario:** Filtrado de términos basándose en los umbrales `min_df` y `max_df`.
    
3. **Vectorización:** Mapeo y recuento de frecuencias para cada documento, resultando en una matriz de dimensión $N_{\text{documentos}} \times N_{\text{vocabulario}}$.

# Entrenamiento y Evaluación del Clasificador de Spam

## 1. Entrenamiento del Modelo con Pipelines y GridSearchCV

Una vez completadas las etapas de preprocesamiento y extracción de características, se procede al entrenamiento del modelo empleando el algoritmo **Multinomial Naive Bayes (`MultinomialNB`)**, el cual es adecuado para clasificación de texto debido a su naturaleza probabilística y su eficiencia procesando matrices dispersas de alta dimensionalidad.

### Encapsulamiento del Flujo mediante Pipelines

Para asegurar un procesamiento consistente y evitar la fuga de datos (_data leakage_), se utiliza la clase `Pipeline` de Scikit-Learn. Esta permite encadenar la etapa de vectorización (`CountVectorizer`) con la etapa de estimación (`MultinomialNB`) en un único objeto unificado.

### Optimización de Hiperparámetros (`GridSearchCV`)

El rendimiento del modelo Multinomial Naive Bayes depende del parámetro de suavizado Laplace/Lidstone ($\alpha$). Este hiperparámetro evita asignaciones de probabilidad nula ($0$) para palabras no observadas en el conjunto de entrenamiento.

Se utiliza `GridSearchCV` con validación cruzada de 5 pliegues ($5$-fold cross-validation) optimizando la métrica $F_1$-score para seleccionar el valor óptimo de $\alpha$:

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Construir la Pipeline combinando vectorización y clasificación
pipeline = Pipeline([
    ("vectorizer", vectorizer),
    ("classifier", MultinomialNB())
])

# Definir la rejilla de hiperparámetros para la búsqueda
param_grid = {
    "classifier__alpha": [0.01, 0.1, 0.15, 0.2, 0.25, 0.5, 0.75, 1.0]
}

# Configurar la búsqueda en rejilla con validación cruzada de 5 pliegues
grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="f1"
)

# Ajustar el modelo con búsqueda de hiperparámetros
grid_search.fit(df["message"], y)

# Extraer el mejor modelo resultante
best_model = grid_search.best_estimator_
print("Mejores parámetros identificados:", grid_search.best_params_)
```

## 2. Evaluación e Inferencia en Nuevos Datos

Para validar la capacidad de generalización del modelo frente a mensajes no vistos previamente, se define un conjunto de prueba con diversos ejemplos de mensajes legítimos y spam.

### Pipeline de Preprocesamiento de Mensajes Nuevos

Cualquier mensaje entrante debe someterse exactamente al mismo tratamiento de texto aplicado durante la fase de entrenamiento:

```python
import numpy as np
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

def preprocess_message(message):
    # Conversión a minúsculas
    message = message.lower()
    # Limpieza de caracteres no permitidos
    message = re.sub(r"[^a-z\s$!]", "", message)
    # Tokenización
    tokens = word_tokenize(message)
    # Eliminación de stop words
    tokens = [word for word in tokens if word not in stop_words]
    # Aplicación de stemming
    tokens = [stemmer.stem(word) for word in tokens]
    # Recomposición del texto
    return " ".join(tokens)
```

### Predicción y Estimación de Probabilidades

Se transforman las cadenas preprocesadas utilizando el vectorizador ajustado dentro de la pipeline y se obtienen tanto la clase predicha como la distribución de probabilidad a posteriori para cada clase:

```python
# Mensajes de prueba para evaluación
new_messages = [
    "Congratulations! You've won a $1000 Walmart gift card. Go to http://bit.ly/1234 to claim now.",
    "Hey, are we still meeting up for lunch today?",
    "Urgent! Your account has been compromised. Verify your details here: www.fakebank.com/verify",
    "Reminder: Your appointment is scheduled for tomorrow at 10am.",
    "FREE entry in a weekly competition to win an iPad. Just text WIN to 80085 now!",
]

# Preprocesamiento de las nuevas entradas
processed_messages = [preprocess_message(msg) for msg in new_messages]

# Transformación de características con el vectorizador ajustado
X_new = best_model.named_steps["vectorizer"].transform(processed_messages)

# Predicción de clases y probabilidades
predictions = best_model.named_steps["classifier"].predict(X_new)
prediction_probabilities = best_model.named_steps["classifier"].predict_proba(X_new)

# Mostrar resultados
for i, msg in enumerate(new_messages):
    prediction = "Spam" if predictions[i] == 1 else "Not-Spam"
    spam_probability = prediction_probabilities[i][1]
    ham_probability = prediction_probabilities[i][0]
    
    print(f"Mensaje: {msg}")
    print(f"Predicción: {prediction}")
    print(f"Probabilidad de Spam: {spam_probability:.2f}")
    print(f"Probabilidad de No-Spam: {ham_probability:.2f}")
    print("-" * 50)
```

## 3. Persistencia de Modelos con `joblib`

En entornos de producción no es viable reentrenar el modelo cada vez que se requiere realizar una inferencia. La serialización permite guardar el estado interno del modelo entrenado (incluyendo los vocabulario ajustados y las probabilidades aprendidas) en un archivo binario.

La librería `joblib` ofrece una alta eficiencia en la serialización y deserialización de objetos de Python con estructuras numéricas complejas (como arreglos NumPy y objetos de Scikit-Learn).

### Guardado del Modelo (`joblib.dump`)

```python
import joblib

# Guardar el modelo entrenado en disco
model_filename = 'spam_detection_model.joblib'
joblib.dump(best_model, model_filename)

print(f"Modelo guardado exitosamente en: {model_filename}")
```

### Carga y Reutilización del Modelo (`joblib.load`)

```python
# Cargar el modelo persistido desde el archivo
loaded_model = joblib.load(model_filename)

# Aplicar preprocesamiento a nuevos datos
new_data_processed = [preprocess_message(msg) for msg in new_messages]

# Inferencia directa con el modelo cargado
predictions = loaded_model.predict(new_data_processed)
```

El uso de `joblib` garantiza despliegues ágiles en producción, reduciendo el tiempo de inicialización y eliminando el costo computacional del entrenamiento en cada reinicio del servicio.
